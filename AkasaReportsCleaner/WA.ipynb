```python
import pandas as pd

# Load CSV
df = pd.read_excel('./csvFiles/WA.xlsx')

# Select columns from 10th to last
start_idx = df.columns.get_loc('Total Sent')
df_cleaned = df[~(df.iloc[:, start_idx:] == 0).all(axis=1)]
df_droppped = df.loc[:, ~df.columns.str.contains('goal 2|goal 3', case=False)]

df_droppped.describe()
df_droppped.head()
```

```python
WAresult = df_droppped.groupby('Tag Category: Uncategorized')[df_droppped.columns].sum()
WAresult.head()
```

```python
WaGroupbyView= df_droppped.groupby('Tag Category: Uncategorized')[df_droppped.columns[df_droppped.columns.get_loc('Total Sent'):]].sum()
print(WaGroupbyView)
```

```python
WATotalsent1 =WaGroupbyView.iloc[0]['Total Sent']
print(int(WATotalsent1))

WATotaldelivered1= WaGroupbyView.iloc[0]['Total Delivered']
print(int(WATotaldelivered1))

WATotalRead1= WaGroupbyView.iloc[0]['Total Read']
print(int(WATotalRead1))

WATotalclicks1=WaGroupbyView.iloc[0]['Total clicks']
print(int(WATotalclicks1))

WAViewconvertedUsers1 = WaGroupbyView.iloc[0]['Goal 1 View Through Converted Users']
print(int(WAViewconvertedUsers1))

WAClickconvertedUsers1 = WaGroupbyView.iloc[0]['Goal 1 Click Through Converted Users']
print(int(WAClickconvertedUsers1))


WAClickconvertedEvents1 =  WaGroupbyView.iloc[0]['Goal 1 Click Through Conversion Events']
print(int(WAClickconvertedEvents1))

WAClickRevenue1 = WaGroupbyView.iloc[0]['Goal 1 Click Through Total Revenue']
print(int(WAClickRevenue1))

Wa_Avg_value1 = WAClickRevenue1/WAClickconvertedEvents1
print(int(Wa_Avg_value1))

Wa_cost1 = WATotalsent1*0.91
print(int(Wa_cost1))
```
