```python
import pandas as pd

# Load CSV
dfApp = pd.read_excel('./csvFiles/push.xlsx')

# Select columns from 10th to last
start_idxApp = dfApp.columns.get_loc('Android Sent')
df_cleanedApp = dfApp[~(dfApp.iloc[:, start_idxApp:] == 0).all(axis=1)]
df_dropppedApp = dfApp.loc[:, ~dfApp.columns.str.contains('goal 2|goal 3', case=False)]
AppGroupbyView= df_dropppedApp.groupby('Tag Category: Uncategorized')[df_dropppedApp.columns[df_dropppedApp.columns.get_loc('Android Sent'):]].sum()
dataApppush = []

for campaign, row in AppGroupbyView.iterrows():
    print(f"{campaign}:")
    
    temp = {'campaign': campaign}
    
    temp['Total Sent'] = int(row['Android Sent'] + row['Ios Sent'])
    temp['Total Impressions'] = int(row['Android Impressions'] + row['Ios Impressions'])
    temp['Total Clicks'] = int(row['Android Clicks'] + row['Ios Clicks'])
    
    temp['Goal 1 View Through Converted Users'] = int(
        row['Goal 1 View Through Converted Users Android'] + 
        row['Goal 1 View Through Converted Users Ios']
    )
    
    temp['Goal 1 Click Through Converted Users'] = int(
        row['Goal 1 Click Through Converted Users Android'] + 
        row['Goal 1 Click Through Converted Users Ios']
    )
    
    temp['Goal 1 Click Through Conversion Events'] = int(
        row['Goal 1 Click Through Conversion Events Android'] + 
        row['Goal 1 Click Through Conversion Events Ios']
    )
    
    temp['Goal 1 Click Through Total Revenue'] = int(
        row['Goal 1 Click Through Total Revenue Android'] + 
        row['Goal 1 Click Through Total Revenue Ios']
    )

    # Print nicely
    for k, v in temp.items():
        if k != 'campaign':
            print(f"  {k}: {v}")
    
    print()
    dataApppush.append(temp)
```
