```python
import pandas as pd

# Load CSV
dfWeb = pd.read_excel('./csvFiles/push.xlsx')

# Select columns from 10th to last
start_idxWeb = dfWeb.columns.get_loc('Android Sent')
df_cleaned = dfWeb[~(dfWeb.iloc[:, start_idxWeb:] == 0).all(axis=1)]
df_droppped = dfWeb.loc[:, ~dfWeb.columns.str.contains('goal 2|goal 3', case=False)]
WebGroupbyView= df_droppped.groupby('Tag Category: Uncategorized')[df_droppped.columns[df_droppped.columns.get_loc('Android Sent'):]].sum()
keep_cols = ['Web Sent', 'Web Impressions', 'Web Clicks', 'Goal 1 View Through Converted Users Web','Goal 1 Click Through Converted Users Web','Goal 1 Click Through Conversion Events Web','Goal 1 Click Through Total Revenue Web']
dataWeb = []
for campaign, row in WebGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}
    for metric in keep_cols:
        print(f"  {metric}: {int(row[metric])}")
        temp[metric] = int(row[metric])
    print()
    dataWeb.append(temp)


```
