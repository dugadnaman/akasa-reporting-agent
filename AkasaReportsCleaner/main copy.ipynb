```python
import pandas as pd

df = pd.read_excel('./csvFiles/email.xls')
start_idx = df.columns.get_loc('Total Sent')
df_cleaned = df[~(df.iloc[:,start_idx:] == 0).all(axis=1)]
df_droppped = df.loc[:, ~df.columns.str.contains('goal 2|goal 3', case=False)]
df_droppped.describe()
df_droppped.head()
emailGroupbyView = df_droppped.groupby('Tag Category: Uncategorized')[df_droppped.columns[df_droppped.columns.get_loc('Total Sent'):]].sum()
print(emailGroupbyView)
keep_cols = ['Total Sent', 'Total Open', 'Total clicks', 'Goal 1 View Through Converted Users','Goal 1 Click Through Converted Users','Goal 1 Click Through Conversion Events','Goal 1 Click Through Total Revenue']
dataEmail = []
for campaign, row in emailGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}
    for metric in keep_cols:
        print(f"  {metric}: {int(row[metric])}")
        temp[metric] = int(row[metric])
    print()
    dataEmail.append(temp)
# Cell 2 — WhatsApp:
dfWA = pd.read_excel('./csvFiles/WA.xls')
start_idxWA = dfWA.columns.get_loc('Total Sent')
df_cleanedWA = dfWA[~(dfWA.iloc[:, start_idxWA:] == 0).all(axis=1)]
df_dropppedWA = dfWA.loc[:, ~dfWA.columns.str.contains('goal 2|goal 3', case=False)]
WaGroupbyView= df_dropppedWA.groupby('Tag Category: Uncategorized')[df_dropppedWA.columns[df_dropppedWA.columns.get_loc('Total Sent'):]].sum()
print(WaGroupbyView)
keep_colsWA = ['Total Sent','Total Delivered' ,'Total Read' ,'Total clicks', 'Goal 1 View Through Converted Users','Goal 1 Click Through Converted Users','Goal 1 Click Through Conversion Events','Goal 1 Click Through Total Revenue']
dataWA = []
for campaign, row in WaGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}
    for metric in keep_colsWA:
        print(f"  {metric}: {int(row[metric])}")
        temp[metric] = int(row[metric])
    print()
    dataWA.append(temp)
# Cell 3 — App Push (Android + iOS):
dfApp = pd.read_excel('./csvFiles/push.xls')

start_idxApp = dfApp.columns.get_loc('Android Sent')
df_cleanedApp = dfApp[~(dfApp.iloc[:, start_idxApp:] == 0).all(axis=1)]
df_dropppedApp = dfApp.loc[:, ~dfApp.columns.str.contains('goal 2|goal 3', case=False)]
AppGroupbyView= df_dropppedApp.groupby('Tag Category: Uncategorized')[df_dropppedApp.columns[df_dropppedApp.columns.get_loc('Android Sent'):]].sum()
dataApppush = []

for campaign, row in AppGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}

    temp['Total Sent'] = int(row['Android Sent'] + row['Ios Sent'])
    temp['Total Impressions'] = int(row['Android Impressions'] + row['Ios Impressions'])
    temp['Total Clicks'] = int(row['Android Clicks'] + row['Ios Clicks'])

    temp['Goal 1 View Through Converted Users'] = int(
        row['Goal 1 View Through Converted Users Android'] +
        row['Goal 1 View Through Converted Users Ios']
    )
    temp['Goal 1 Click Through Converted Users'] = int(
        row['Goal 1 Click Through Converted Users Android'] +
        row['Goal 1 Click Through Converted Users Ios']
    )
    temp['Goal 1 Click Through Conversion Events'] = int(
        row['Goal 1 Click Through Conversion Events Android'] +
        row['Goal 1 Click Through Conversion Events Ios']
    )
    temp['Goal 1 Click Through Total Revenue'] = int(
        row['Goal 1 Click Through Total Revenue Android'] +
        row['Goal 1 Click Through Total Revenue Ios']
    )

    for k, v in temp.items():
        if k != 'campaign':
            print(f"  {k}: {v}")
    print()
    dataApppush.append(temp)
# Cell 4 — Web Push:
dfWeb = pd.read_excel('./csvFiles/push.xls')

dfWeb = dfWeb.loc[:, ~dfWeb.columns.str.contains('goal 2|goal 3', case=False)]

start_idx_web = dfWeb.columns.get_loc('Web Sent')

dfWeb_cleaned = dfWeb[~(dfWeb.iloc[:, start_idx_web:] == 0).all(axis=1)]

WebGroupbyView = dfWeb_cleaned.groupby('Tag Category: Uncategorized')[
    dfWeb_cleaned.columns[start_idx_web:]
].sum()

web_mapping = {
    'Web Sent': 'Total Sent',
    'Web Impressions': 'Total Delivered',
    'Web Clicks': 'Total Clicks',
    'Goal 1 View Through Converted Users Web': 'Goal 1 View Through Converted Users',
    'Goal 1 Click Through Converted Users Web': 'Goal 1 Click Through Converted Users',
    'Goal 1 Click Through Conversion Events Web': 'Goal 1 Click Through Conversion Events',
    'Goal 1 Click Through Total Revenue Web': 'Goal 1 Click Through Total Revenue'
}

dataWeb = []
for campaign, row in WebGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}
    for old_col, new_col in web_mapping.items():
        value = int(row[old_col])
        print(f"  {new_col}: {value}")
        temp[new_col] = value
    print()
    dataWeb.append(temp)
# Cell 5 — Overall (with Grand Total):
dfOver = pd.read_excel('./csvFiles/overall.xls')

start_idxOver = dfOver.columns.get_loc('Trips Started Total users')
df_cleanedOver = dfOver[~(dfOver.iloc[:,start_idxOver:] == 0).all(axis=1)]
df_dropppedOver = dfOver.loc[:, ~dfOver.columns.str.contains('goal 3', case=False)]
OverallGroupbyView = df_dropppedOver.groupby('Tag Category: Uncategorized')[df_dropppedOver.columns[df_dropppedOver.columns.get_loc('Trips Started Total users'):]].sum()
print(OverallGroupbyView)

keep_cols = ['Trips Started Total users', 'Trips engaged', 'Goal 1 View Through Total Conversion users (CVR)', 'View Through Revenue (Primary Goal)']
dataOverAll = []
for campaign, row in OverallGroupbyView.iterrows():
    print(f"{campaign}:")
    temp = {'campaign': campaign}
    for metric in keep_cols:
        print(f"  {metric}: {int(row[metric])}")
        temp[metric] = int(row[metric])
    print()
    dataOverAll.append(temp)

# Add grand total row
total_row = {'campaign': 'Grand Total'}
for metric in keep_cols:
    total_row[metric] = int(OverallGroupbyView[metric].sum())
dataOverAll.append(total_row)
print('Grand Total:')
for metric in keep_cols:
    print(f"  {metric}: {total_row[metric]}")
# Cell 6 — Consolidation & Export:
all_data = {
    'Email': dataEmail,
    'WhatsApp': dataWA,
    'App Push': dataApppush,
    'Web': dataWeb,
    'overAll': dataOverAll
}

dfs = []
for channel, data in all_data.items():
    df_temp = pd.DataFrame(data)
    df_temp['Channel'] = channel
    dfs.append(df_temp)

final_df = pd.concat(dfs, ignore_index=True)

grouped_df = final_df.groupby(['campaign', 'Channel'], as_index=False).sum()

grouped_df.to_excel("final_campaign_report.xlsx", index=False)
```
